```python
#Count occurences of each word
text=input("Enter a line of text:")
words=text.split()
count={}
for i in words:
    count[i]=count.get(i,0)+1
print("Word occurences:")
for i,j in count.items():
    print(i,":",j)
```

```python
#Replace values greater than 100
numbers=list(map(int,input("Enter integer seperated by spaces:").split()))
for i in range(len(numbers)):
    if numbers[i]>100:
        numbers[i]="over"
print("Result:",numbers)
```

```python
#Count 'a' in first names
names=input("Enter first names seperated by spaces:").split()
count=0
for n in names:
    count+=n.lower().count('a')
print("Number of 'a'occurences:",count)
```

```python
#Find GCD
a=int(input("Enter first number:"))
b=int(input("Ente second number:"))
while b!=0:
    a,b=b,a%b
print("GCD:",a)
```

```python
#Remove Even numbers
numbers=list(map(int,input("Enter integers seperated by spaces:").split()))
result=[]
for num in numbers:
    if num%2!=0:
        result.append(num)
print("List after removing even numbers:",result)
```

```python
#Find Factorial of a number
n=int(input("Enter a number:"))
fact=1
for i in range(1,n+1):
    fact=fact*i
print("Factorial:",fact)
```

```python
#Fibonacci Series of n terms
n=int(input("Enter the number of terms:"))
a=0
b=1
for i in range(n):
    print(a,end=" ")
    a,b=b,a+b
```

```python
#Sum of List items
numbers=list(map(int,input("Enter numbers seperated by spaces:").split()))
total=0
for num in numbers:
    total=total+num
print("Sum:",total)
```

```python
#Even Digit Perfect Squares
import math
start=int(input("Enter starting range:"))
end=int(input("Enter ending range:"))
result=[]
for num in range(start,end+1):
    if 1000<=num<=9999:
        if int(math.sqrt(num))**2==num:
            if all(int(digit)%2==0 for digit in str(num)):
                result.append(num)
print("Result:",result)
```

```python
#Character Frequency
text=input("Enter a string:")
count={}
for char in text:
    count[char]=count.get(char,0)+1
print("Character Frequency:")
for char,frequency in count.items():
    print(char," ",frequency)
```

```python
#Compare Two Lists
l1=list(map(int,input("Enter first list:").split()))
l2=list(map(int,input("Enter second list:").split()))
#check same length
if len(l1)==len(l2):
    print("Both lists have the same length")
else:
    print("Both lists do not have the same length")
#check same sum
if sum(l1)==sum(l2):
    print("Both lists have the same sum")
else:
    print("Both lists do not have the same sum")
#check common value
if set(l1)&set(l2):
    print("Both lists have a common value")
else:
    print("No common value")
```

```python
#Display pyramid
n=int(input("Enter number of steps:"))
for i in range(1,n+1):
    for j in range(1,i+1):
        print(i*j,end=" ")
    print()
```
