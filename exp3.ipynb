```python
# Given string to a new string where first and last chars have been exchanged
s=input("Enter  a string:")
new_s=s[-1]+s[1:-1]+s[0]
print("New String=",new_s)
print(s[::-1])#reversing string
print(s[0:5:2])#prints from 0 to 4 by skipping one char
```

```python
# String made of the first 2 and last 2 chars from a given string
s=input("Enter a string:")
fl_string=s[0:2]+s[-2:]
print("New String=",fl_string)
```

```python
# Create a single string seperated with space from two strings by swapping the char at position 1 
s1="hello"
s2="world"
a=s2[0]+s1[1:]+" "+s1[0]+s2[1:]
print("Before Swapping:",s1,s2)
print("swapped string:",a)
```

```python
#To get a string from a given string where all occurences of its first char have been changed to $, expect first char itself
s=input("Enter  a string:")
f=s[0]
rep=f+s[1:].replace(f,'$')
print("Normal string:",s)
print("Replaced by '$':",rep)
```
