```python
#Python program to add "ing" to the end of a given string. If the string already ends with "ing", add"ly" instead.
s=input("Enter a string:")
if s.endswith("ing"):#we can also give 's[-3:]=="ing"'
    result=s+"ly"
else:
    result=s+"ing"
print("String:",s)
print("Result=",result)
```

```python
#Python program to check the given string is palindrome or not
s=input("Enter a string:")
print("String:",s)
print("Reversed String:",s[::-1])
if s==s[::-1]:
    print("its a palindrome")
else:
    print("its not a palindrome")
```
