```python
# python program to accept 5 students name and roll number and map it into dictionary
students={}

name1=input("Enter student name:")
roll1=int(input("Enter roll number:"))
students[roll1]=name1

name2=input("Enter student name:")
roll2=int(input("Enter roll number:"))
students[roll2]=name2

name3=input("Enter student name:")
roll3=int(input("Enter roll number:"))
students[roll3]=name3

name4=input("Enter student name:")
roll4=int(input("Enter roll number:"))
students[roll4]=name4

name5=input("Enter student name:")
roll5=int(input("Enter roll number:"))
students[roll5]=name5

ascending=dict(sorted(students.items()))
descending=dict(sorted(students.items(),reverse=True))
print("\nOriginal dictionary:",students)
print("\nAscending Order:",ascending)
print("\nDescending Order:",descending)
```

```python
#Python program to accept 2 dictionaries (dic1 contains 3 students name and roll no.s,and dic2 contains 2 students names and roll no.s)Merge 2 dictionaries and display it[use update()]
dic1={} 
dic2={}
print("Enter 3 students for Dictionary 1:")

name1=input("Enter student name:")
roll1=int(input("Enter roll number:"))
dic1[name1]=roll1

name2=input("Enter student name:")
roll2=int(input("Enter roll number:"))
dic1[name2]=roll2

name3=input("Enter student name:")
roll3=int(input("Enter roll number:"))
dic1[name3]=roll3

print("\nEnter 2 students for Dictionary 2:")

name4=input("Enter student name:")
roll4=int(input("Enter roll number:"))
dic2[name4]=roll4

name5=input("Enter student name:")
roll5=int(input("Enter roll number:"))
dic2[name5]=roll5

print("\nDictionary 1:",dic1)
print("\nDictionary 2:",dic2)
dic1.update(dic2)
print("\nMerged Dictionary:",dic1)
```
