```python
#write a python program to accept 5 students name and roll number and map it into dictionary
students={}
for i in range(5):
    name=input("Enter student name:")
    roll=int(input("Enter roll number:"))
    students[roll]=name
    ascending=dict(sorted(students.items()))
    descending=dict(sorted(students.items(),reverse=True))
    print("Original dictionary:",students)
    print("Ascending Order:",ascending)
    print("Descending Order:",descending)
```
