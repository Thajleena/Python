```python
#Generating a list of positive numbers using list comprehension
a=list(map(int,input("Enter numbers seperated by space:").split()))
positive=[n for n in a if n>0]
print("Positive numbers:",positive)
```

```python
#Finding the squares of N numbers using list comprehension
n=int(input("Enter N:"))
a=[int(input("Enter number:")) for i in range(n)]
square=[x*x for x in a]
print("Squares:",square)
```

```python
#Extracting Vowels from a word using list comprehension
word=input("Enter a word:")
vowels=[ch for ch in word if ch in "aeiouAEIOU"]
print("Vowels:",vowels)
```

```python
#Finding the Ordinal values of characters in a word using list comprehension
word=input("Enter a word:")
values=[ord(ch) for ch in word]
print("Ordinal values:",values)
```
