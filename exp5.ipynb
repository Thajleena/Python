```python
#write a python program to get 3rd element from last of a tuple
numbers=input("Enter tuple elements seperated by commas:")
t=tuple(map(int, numbers.split(',')))
print("Tuple:",t)
print("Enter 3rd element:",t[2])
print("Enter 3rd element from the last:",t[-3])
```

```python
#write a python program which accepts sequence of comma seperated numbers from user and generate a tuple with those numbersand find averageof all items in tuple
numbers=input("Enter numbers seperated by commas:")
t=tuple(map(float,numbers.split(',')))
average=sum(t)/len(t)
print("Tuple:",t)
print("Average:",average)
```
