```python
class Rectangle:
    def get(self,l,b):
        self.l=l
        self.b=b
    def area(self,l,b):
        return l*b
    def perimeter(self,l,b):
        return 2*(l+b)
r1=Rectangle()
r2=Rectangle()
l1=int(input("Enter length of rectangle 1:"))
b1=int(input("Enter breadth of rectangle 1:"))
l2=int(input("Enter length of rectangle 2:"))
b2=int(input("Enter breadth of rectangle 2:"))
p=r1.area(l1,b1)
q=r2.area(l2,b2)
s=r1.perimeter(l1,b1)
t=r2.perimeter(l2,b2)
print("\nRectangle 1:")
print("Area of rectangle 1:",p)
print("Perimeter of rectangle 1:",s)
print("\nRectangle 2:")
print("Area of rectangle 2:",q)
print("Perimeter of rectangle 2:",t)
if p>q:
    print("\nRectangle 1 is greater")
elif p<q:
    print("\nRectangle 2 is greater")
else:
    print("\nRectangle are equal")
```

```python
class BankAccount:
    def __init__(self,name,accno,acctype,balance):
        self.name=name
        self.accno=accno
        self.acctype=acctype
        self.balance=balance
    def deposit(self,amount):
        print("Initial balance:",self.balance)
        self.balance+=amount
        print("Updated Balance:",self.balance)
    def withdraw(self):
        amount=int(input("Enter withdrawal amount:"))
        if amount>self.balance:
            print("Not enough balance!!")
        else:
            self.balance-=amount
            print("updated balance:",self.balance)
    def acc_info(self):
        print("Account Holder:",self.name)
        print("Accountnumber:",self.accno)
        print("Account type:",self.acctype)
        print("Balance:",self.balance)
name=input("Enter Name:")
accno=int(input("Enter account number:"))
acctype=input("Enter account type:")
balance=int(input("Enter initial balance:"))
holder=BankAccount(name,accno,acctype,balance)
while True:
    print("\n1.Deposit")
    print("2.Withdraw")
    print("3.Account Information")
    print("4.Exit")
    choice=int(input("Enter your choice:"))
    if choice==1:
        amount=int(input("Enter deposit amount:"))
        holder.deposit(amount)
    elif choice==2:
        holder.withdraw()
    elif choice==3:
        holder.acc_info()
    elif choice==4:
        break
    else:
        print("Invalid choice")
    
    
```
