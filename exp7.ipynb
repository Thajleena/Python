```python
#write a python program to printout all colors from colorset 1 not contain in colorset 2
color_set1=set(input("Enter colors in color set 1:").split(','))
color_set2=set(input("Enter colors in color set 2:").split(','))
print("color set 1:",color_set1)
print("color set 2:",color_set2)
result=color_set1-color_set2
print("Colors in colorset 1 but not contain in colorset 2:",result)
```

```python
#write a python program to printout all colors from colorset 1 and colorset 2
color_set1=set(input("Enter colors in color set 1:").split(','))
color_set2=set(input("Enter colors in color set 2:").split(','))
print("color set 1:",color_set1)
print("color set 2:",color_set2)
result=color_set1|color_set2
print("Enter all colors from colorset 1 and colorset 2:",result)
```
